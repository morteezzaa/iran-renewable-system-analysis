In this notebook, first the model created by model.ipynb is imported and then the visualisations are created.

```python
def _weights(net):
    """Snapshot weights for objective aggregation (hours per snapshot)."""
    w = net.snapshot_weightings["objective"]
    if isinstance(w, pd.Series):
        return w.reindex(net.snapshots).fillna(1.0)
    return pd.Series(1.0, index=net.snapshots)

def _energy_mwh(net, p_df):
    """Convert power time series (MW) to energy (MWh) using snapshot weights."""
    w = _weights(net)
    return p_df.mul(w, axis=0).sum(axis=0)  # MWh per asset/column

def _carrier_series(net, df, carrier_index):
    """Group columns of df by carrier_index (a Series mapping col -> carrier)."""
    # df: snapshots x assets
    return df.groupby(carrier_index, axis=1).sum()

def _get_opt_capacity(net, component):
    """Return p_nom_opt if exists, else p_nom."""
    comp = getattr(net, component)
    if "p_nom_opt" in comp.columns:
        return comp["p_nom_opt"].fillna(comp["p_nom"])
    return comp["p_nom"]

def _safe_plot_series(s, title, ylabel):
    ax = s.sort_values(ascending=False).plot(kind="bar", figsize=(10,4))
    ax.set_title(title)
    ax.set_ylabel(ylabel)
    plt.tight_layout()
    plt.show()

def _ensure_prices(net):
    """Return bus marginal prices time series (snapshots x buses)."""
    if hasattr(net, "buses_t") and "marginal_price" in net.buses_t:
        return net.buses_t.marginal_price
    raise ValueError("No bus prices found (buses_t.marginal_price). Did the optimisation compute prices?")
```

### TOTAL SYSTEM COST SPLIT BY TECHNOLOGY

```python
def plot_cost_split_by_tech(net, title_prefix=""):
    w = _weights(net)

    # --- Generators: capex (for extendable) and opex (dispatch * marginal_cost) ---
    gen = net.generators.copy()
    gen_cap = _get_opt_capacity(net, "generators")  # MW
    gen_car = gen["carrier"]

    # Capex: capital_cost [€/MW/year] * p_nom_opt [MW]
    gen_capex = (gen.get("capital_cost", 0.0).fillna(0.0) * gen_cap).groupby(gen_car).sum()

    # Opex: sum_t p(t) [MW]*w[h]*marginal_cost[€/MWh]
    if hasattr(net, "generators_t") and "p" in net.generators_t:
        gen_energy = _energy_mwh(net, net.generators_t.p)  # MWh per generator
        gen_opex = (gen_energy * gen.get("marginal_cost", 0.0).fillna(0.0)).groupby(gen_car).sum()
    else:
        gen_opex = pd.Series(dtype=float)
        # --- StorageUnits: capex and opex (dispatch * marginal_cost) ---
    su_capex = pd.Series(0.0, index=pd.Index([], name="carrier"))
    su_opex  = pd.Series(0.0, index=pd.Index([], name="carrier"))
    if len(net.storage_units) > 0:
        su = net.storage_units.copy()
        su_cap = _get_opt_capacity(net, "storage_units")  # MW
        su_car = su["carrier"]

        su_capex = (su.get("capital_cost", 0.0).fillna(0.0) * su_cap).groupby(su_car).sum()

        if hasattr(net, "storage_units_t") and "p" in net.storage_units_t:
            su_energy = _energy_mwh(net, net.storage_units_t.p)  # MWh per storage unit (net dispatch)
            su_opex = (su_energy.abs() * su.get("marginal_cost", 0.0)).fillna.sum()
                       
                        # --- Links: treat as transmission investment (capex) + opex (if marginal_cost exists) ---
    link_capex = pd.Series(dtype=float)
    link_opex  = pd.Series(dtype=float)
    if len(net.links) > 0:
        links = net.links.copy()
        link_cap = _get_opt_capacity(net, "links")  # MW
        link_car = links.get("carrier", pd.Series("link", index=links.index))

        link_capex = (links.get("capital_cost", 0.0).fillna(0.0) * link_cap).groupby(link_car).sum()

        if hasattr(net, "links_t") and "p0" in net.links_t:
            link_energy = _energy_mwh(net, net.links_t.p0.abs())  # MWh per link
            link_opex = (link_energy * links.get("marginal_cost", 0.0).fillna(0.0)).groupby(link_car).sum()

    # Combine into a DataFrame
    all_carriers = pd.Index(sorted(set(gen_capex.index) | set(gen_opex.index) | set(su_capex.index) |
                                    set(su_opex.index) | set(link_capex.index) | set(link_opex.index)))
    capex = (gen_capex.reindex(all_carriers, fill_value=0.0)
             + su_capex.reindex(all_carriers, fill_value=0.0)
             + link_capex.reindex(all_carriers, fill_value=0.0))
    opex  = (gen_opex.reindex(all_carriers, fill_value=0.0)
             + su_opex.reindex(all_carriers, fill_value=0.0)
             + link_opex.reindex(all_carriers, fill_value=0.0))

    cost_df = pd.DataFrame({"Capex (€/y)": capex, "Opex (€/y)": opex})
    cost_df = cost_df.loc[cost_df.sum(axis=1).sort_values(ascending=False).index]

    ax = cost_df.plot(kind="bar", stacked=True, figsize=(11,5))
    ax.set_title(f"{title_prefix} Total system cost split by technology")
    ax.set_ylabel("€/year")
    plt.tight_layout()
    plt.show()

    return cost_df

    
```

### Capacities built per technology (MW)

```python
def plot_capacities_built(net, title_prefix=""):
    gen_cap = _get_opt_capacity(net, "generators")
    cap_by_car = gen_cap.groupby(net.generators["carrier"]).sum()

    # include storage power capacities too (optional)
    if len(net.storage_units) > 0:
        su_cap = _get_opt_capacity(net, "storage_units")
        cap_by_car = cap_by_car.add(su_cap.groupby(net.storage_units["carrier"]).sum(), fill_value=0.0)

    _safe_plot_series(cap_by_car, f"{title_prefix} Capacities built per technology", "MW")
    return cap_by_car
```

###  Electricity mix (energy by carrier, MWh)

```python
def plot_electricity_mix(net, title_prefix=""):
    if not (hasattr(net, "generators_t") and "p" in net.generators_t):
        raise ValueError("No generator dispatch found: generators_t.p")

    gen_p = net.generators_t.p
    gen_by_car = _carrier_series(net, gen_p, net.generators["carrier"])
    gen_energy_by_car = _energy_mwh(net, gen_by_car)  # MWh per carrier

    _safe_plot_series(gen_energy_by_car, f"{title_prefix} Electricity mix (generation)", "MWh")
    return gen_energy_by_car
```

### CO2 shadow price (dual variable)

```python
def plot_co2_shadow_price(net, title_prefix="", constraint_name="CO2Limit"):
    if len(net.global_constraints) == 0 or constraint_name not in net.global_constraints.index:
        print(f"No global constraint '{constraint_name}' found in this network.")
        return None

    mu = net.global_constraints.loc[constraint_name, "mu"]
    # mu is the shadow price of the constraint in €/ (unit of constraint)
    # If your carrier_attribute is 'co2_emissions' (tCO2/MWh) then mu is often ~€/tCO2.
    plt.figure(figsize=(6,3))
    plt.bar([f"{constraint_name}"], [mu])
    plt.title(f"{title_prefix} CO2 shadow price (mu)")
    plt.ylabel("€ / (constraint unit)")
    plt.tight_layout()
    plt.show()
    return mu
```

### Price duration curves (system and per region)

```python
def plot_price_duration_curves(net, title_prefix="", top_regions=6):
    prices = _ensure_prices(net)  # snapshots x buses

    # System price (load-weighted average) if loads exist
    if len(net.loads) > 0 and hasattr(net, "loads_t") and "p_set" in net.loads_t:
        load = net.loads_t.p_set
        load_by_bus = load.groupby(net.loads.bus, axis=1).sum()
        # Align with prices buses
        common = load_by_bus.columns.intersection(prices.columns)
        load_by_bus = load_by_bus[common]
        prices_use = prices[common]
        system_price = (prices_use * load_by_bus).sum(axis=1) / load_by_bus.sum(axis=1).replace(0, np.nan)
        system_price = system_price.fillna(prices_use.mean(axis=1))
    else:
        system_price = prices.mean(axis=1)

    plt.figure(figsize=(10,5))
    system_sorted = np.sort(system_price.values)[::-1]
    plt.plot(system_sorted, label="System (avg)")

    # Plot a few regions (highest average price)
    avg_bus_price = prices.mean(axis=0).sort_values(ascending=False)
    for b in avg_bus_price.head(top_regions).index:
        plt.plot(np.sort(prices[b].values)[::-1], alpha=0.7, label=b)

    plt.title(f"{title_prefix} Price duration curves")
    plt.ylabel("€/MWh")
    plt.xlabel("Hours (sorted)")
    plt.legend(ncol=2, fontsize=8)
    plt.tight_layout()
    plt.show()
```

### Average electricity prices per region (€/MWh)

```python
def plot_avg_prices_per_region(net, title_prefix=""):
    prices = _ensure_prices(net)
    avg = prices.mean(axis=0).sort_values(ascending=False)
    _safe_plot_series(avg, f"{title_prefix} Average electricity prices per region", "€/MWh")
    return avg
```

### Curtailment rate (VRE)

```python
def plot_curtailment(net, title_prefix="", vre_carriers=("solar", "onwind", "offwind")):
    if not (hasattr(net, "generators_t") and "p" in net.generators_t):
        raise ValueError("No generator dispatch found: generators_t.p")

    gens = net.generators.copy()
    vre = gens[gens["carrier"].isin(vre_carriers)].index
    if len(vre) == 0:
        print("No VRE generators found with carriers:", vre_carriers)
        return None

    p = net.generators_t.p[vre]  # MW
    cap = _get_opt_capacity(net, "generators").reindex(vre)  # MW

    # Available = p_max_pu * p_nom_opt
    if hasattr(net, "generators_t") and "p_max_pu" in net.generators_t:
        pmaxpu = net.generators_t.p_max_pu[vre]  # 0..1
    else:
        # fallback: use static p_max_pu if time series absent
        pmaxpu = pd.DataFrame(1.0, index=net.snapshots, columns=vre)

    available = pmaxpu.mul(cap, axis=1)  # MW
    w = _weights(net)
    avail_mwh = available.mul(w, axis=0).sum(axis=0)
    gen_mwh   = p.mul(w, axis=0).sum(axis=0)
    curtailed_mwh = (avail_mwh - gen_mwh).clip(lower=0.0)

    curtail_rate = (curtailed_mwh / avail_mwh.replace(0, np.nan)).fillna(0.0) * 100.0  # %

    # By carrier (weighted)
    curtail_by_car = curtail_rate.groupby(gens.loc[vre, "carrier"]).mean().sort_values(ascending=False)

    _safe_plot_series(curtail_by_car, f"{title_prefix} Curtailment rate by VRE technology", "%")
    return curtail_rate, curtail_by_car
```

### Storage filling levels (SOC) throughout the year

```python
def plot_storage_soc(net, title_prefix="", carriers=("battery", "H2"), max_units=6):
    if len(net.storage_units) == 0 or not (hasattr(net, "storage_units_t") and "state_of_charge" in net.storage_units_t):
        print("No StorageUnit SOC data found.")
        return

    su = net.storage_units.copy()
    soc = net.storage_units_t.state_of_charge  # MWh

    # pick a few units per carrier
    plt.figure(figsize=(11,5))
    plotted = 0
    for c in carriers:
        units = su[su["carrier"] == c].index
        for u in units[:max_units]:
            plt.plot(soc[u].values, label=f"{c}:{u}", alpha=0.7)
            plotted += 1
            if plotted >= max_units:
                break
        if plotted >= max_units:
            break
    plt.title(f"{title_prefix} Storage filling levels (SOC)")
    plt.ylabel("MWh")
    plt.xlabel("Snapshots")
    plt.legend(fontsize=7, ncol=2)
    plt.tight_layout()
    plt.show()
```

### Operation of the system for a selected week

```python
def plot_operation_week(net, start="2013-01-01", days=7, title_prefix=""):
    # pick window
    start = pd.Timestamp(start)
    end = start + pd.Timedelta(days=days)
    sns = net.snapshots[(net.snapshots >= start) & (net.snapshots < end)]
    if len(sns) == 0:
        # fallback: first week in snapshots
        sns = net.snapshots[:days*8]  # for 3-hourly, 8 steps/day

    # generation by carrier
    gen_by_car = _carrier_series(net, net.generators_t.p.loc[sns], net.generators["carrier"])

    # load total
    if hasattr(net, "loads_t") and "p_set" in net.loads_t:
        load_total = net.loads_t.p_set.loc[sns].sum(axis=1)
    else:
        load_total = None

    ax = gen_by_car.plot(kind="area", stacked=True, figsize=(12,5))
    if load_total is not None:
        ax.plot(load_total.index, load_total.values, linewidth=2, label="Load", color="black")

    ax.set_title(f"{title_prefix} System operation (selected week)")
    ax.set_ylabel("MW")
    ax.legend(ncol=3, fontsize=8)
    plt.tight_layout()
    plt.show()
```
